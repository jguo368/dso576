# DHCS Certified Eligibles by Primary Language

This notebook reruns the reproducible cleaner from the unchanged DHCS source, then inspects the whole-file QA summary, difficult-case sample, and five record checks. The primary map measure is a reported county-month-language count. Suppressed values stay unknown; the statewide and invalid-county rows are retained in the output but are not county-map records.

Source coverage confirmed by the publisher: January 2010 through August 2026. Recent eligibility data is preliminary and subject to change for one year.

```python
from pathlib import Path
import sys

PROJECT_DIR = Path.cwd()
if PROJECT_DIR.name != "project":
    PROJECT_DIR = PROJECT_DIR / "project"
sys.path.insert(0, str(PROJECT_DIR))

from clean_language_data import clean_file, DEFAULT_INPUT, DEFAULT_OUTPUT_DIR

summary = clean_file(DEFAULT_INPUT, DEFAULT_OUTPUT_DIR)
print(f"Source rows: {summary['input_rows']:,}")
print(f"Cleaned rows: {summary['output_rows']:,}")
print(f"Month range: {summary['month_min']} to {summary['month_max']}")
print(f"Reported count rows: {summary['reported_count_rows']:,}")
print(f"Suppressed count rows: {summary['suppressed_count_rows']:,}")
print(f"Map counties: {summary['map_county_count']}")
print(f"All record checks passed: {summary['record_checks_passed']}")
```
